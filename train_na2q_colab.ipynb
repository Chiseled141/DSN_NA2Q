```python
# ── 4. Verify GPU ──────────────────────────────────────────────────────────────
import torch
print('CUDA available:', torch.cuda.is_available())
if torch.cuda.is_available():
    print('GPU:', torch.cuda.get_device_name(0))
    print('Memory:', round(torch.cuda.get_device_properties(0).total_memory / 1e9, 1), 'GB')
```

```python
# ── 5. Train Scenario 1 (50k episodes) ────────────────────────────────────────
!python -m na2q.main --mode train --scenario 1
```

```python
# ── 6. Save Scenario 1 results to Drive ───────────────────────────────────────
import shutil

CKPT_DIR = 'na2q/checkpoints'
SAVE_FILES = ['best_model.pt', 'final_model.pt', 'training_history.npz']

saved = []
for fname in SAVE_FILES:
    src = os.path.join(CKPT_DIR, fname)
    if os.path.exists(src):
        shutil.copy(src, f'{DRIVE_DIR}/scenario1/{fname}')
        saved.append(fname)

print('Scenario 1 saved to Drive:', saved)
```

```python
# ── 7. Train Scenario 2 (10k episodes) ────────────────────────────────────────
!python -m na2q.main --mode train --scenario 2
```

```python
# ── 8. Save Scenario 2 results to Drive ───────────────────────────────────────
saved = []
for fname in SAVE_FILES:
    src = os.path.join(CKPT_DIR, fname)
    if os.path.exists(src):
        shutil.copy(src, f'{DRIVE_DIR}/scenario2/{fname}')
        saved.append(fname)

print('Scenario 2 saved to Drive:', saved)
```

```python
# ── 9. Download all results to your machine ────────────────────────────────────
from google.colab import files

for scenario in [1, 2]:
    for fname in SAVE_FILES:
        path = f'{DRIVE_DIR}/scenario{scenario}/{fname}'
        if os.path.exists(path):
            files.download(path)
```
